# 03 Feature Store

Este notebook constrói atributos históricos reutilizáveis a partir das bases cadastral, dos empréstimos e das parcelas.

## Abordagem

Este notebook utiliza funções modulares em `src/feature_store.py` para construir as feature stores de treino e score a partir das bases de cadastro, empréstimos e parcelas.


```python
import sys
from pathlib import Path
sys.path.append(str(Path('..').resolve()))

from src.feature_store import build_feature_store, build_score_feature_store

feature_store = build_feature_store()
score_features = build_score_feature_store()

feature_store.head()
```

```python
from pathlib import Path
output_path = Path('..') / 'data' / 'processed'
output_path.mkdir(parents=True, exist_ok=True)
feature_store.to_parquet(output_path / 'feature_store.parquet', index=False)
score_features.to_parquet(output_path / 'score_feature_store.parquet', index=False)
print('Feature store shape:', feature_store.shape)
print('Score feature store shape:', score_features.shape)
```
