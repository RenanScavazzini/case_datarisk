Transformações finais, encoding e preparação do dataset de modelagem.

## Abordagem

Transformamos as features finais para modelagem, garantindo que variáveis ausentes sejam tratadas e categorias sejam codificadas de forma consistente para treino e score.


```python
import sys
from pathlib import Path
sys.path.append(str(Path('..').resolve()))

import pandas as pd
from src.feature_engineering import prepare_model_dataset

feature_store = pd.read_parquet(Path('..') / 'data' / 'processed' / 'feature_store.parquet')
target_population = pd.read_parquet(Path('..') / 'data' / 'processed' / 'population_target.parquet')
model_source = feature_store.merge(target_population[['id_contrato', 'target']], on='id_contrato', how='left')
model_source = model_source[model_source['target'].notna()].copy()
X, y, processed = prepare_model_dataset(model_source, target_col='target')
print('Model dataset shape:', X.shape)
print('Target share:', y.mean())
```

```python
processed = processed.copy()
processed['target'] = y.values
processed_path = Path('..') / 'data' / 'processed' / 'model_dataset.parquet'
processed.to_parquet(processed_path, index=False)
print('Saved model dataset to', processed_path)
```

## Decisões de transformação

- As variáveis categóricas foram codificadas como valores ordinais.
- Valores numéricos ausentes foram preenchidos com mediana.
- Novas variáveis de alavancagem e de renda foram criadas para capturar o risco relativo da operação.
