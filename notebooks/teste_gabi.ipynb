```python
import random

nomes_gatos = [
    "Luna", "Mingau", "Nina", "Tom", "Mia", "Simba", "Mel", "Theo", "Bela", "Chico",
    "Amora", "Fred", "Pandora", "Pipoca", "Cacau", "Frajola", "Nick", "Paçoca", "Yuki", "Sushi"
]

cores_gatos = [
    "Preto", "Branco", "Cinza", "Caramelo", "Rajado", "Tigrado", "Tricolor", "Laranja", "Creme", "Malhado"
]

dados_gatos = [
    (random.choice(nomes_gatos), random.choice(cores_gatos))
    for _ in range(100)
]

gatos_df = spark.createDataFrame(dados_gatos, ["Nome", "Cor"])
display(gatos_df)
```

```python
print(f"Total de linhas: {gatos_df.count()}")

print("Top 5 nomes mais comuns")
display(
    gatos_df.groupBy("Nome")
    .count()
    .orderBy("count", ascending=False)
    .limit(5)
)

print("Top 5 cores mais comuns")
display(
    gatos_df.groupBy("Cor")
    .count()
    .orderBy("count", ascending=False)
    .limit(5)
)
```
