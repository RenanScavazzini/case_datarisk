## 5. Divisão Train e OOS

Após a aplicação de todas as etapas de Feature Engineering, a base de desenvolvimento será dividida em duas amostras:

- **Train (Training Sample)**: utilizada para treinamento e ajuste dos modelos;
- **OOS (Out Of Sample)**: utilizada para validação do desempenho do modelo em observações que não participaram do treinamento.

A divisão será realizada utilizando amostragem estratificada pela variável alvo (`target`), garantindo que a proporção de bons e maus pagadores permaneça semelhante entre as duas amostras.

Será adotada uma divisão de **80% para Train** e **20% para OOS**, proporção amplamente utilizada em projetos de modelagem preditiva por proporcionar uma quantidade suficiente de observações para treinamento sem comprometer a capacidade de validação do modelo.

Além das bases de desenvolvimento, será mantida separadamente a base **OOT (Out Of Time)**, composta por observações de um período posterior ao utilizado na construção do modelo. Essa base não participa do processo de treinamento e será utilizada para avaliar a estabilidade temporal e a capacidade de generalização do modelo em um cenário mais próximo da produção.

Ao final desta etapa serão apresentadas as quantidades de registros, clientes bons (`good`) e maus (`bad`) nas bases Train, OOS e OOT, permitindo validar a representatividade das amostras geradas.

```python
train_base_final, oos_base = (
    train_test_split(
        train_base,
        test_size=0.20,
        random_state=42,
        stratify=train_base["target"]
    )
)
```

```python
for nome, base in {
    "Train": train_base_final,
    "OOS": oos_base,
    "OOT": oot_base
}.items():

    bad = base["target"].sum()

    good = (
        len(base)
        - bad
    )

    print(
        f"{nome}: "
        f"Good={good:,} "
        f"Bad={bad:,}"
    )
```

## 6. Balanceamento da Base de Treino

A variável alvo apresenta forte desbalanceamento entre clientes adimplentes e inadimplentes.

Para reduzir o impacto desse desbalanceamento durante o treinamento dos modelos, será aplicada a técnica de Random Under Sampling (RUS) exclusivamente na base de treino.

A técnica consiste em reduzir aleatoriamente a quantidade de observações da classe majoritária, tornando a distribuição das classes mais equilibrada.

As bases OOS e OOT não serão balanceadas, preservando a distribuição real observada na população e permitindo uma avaliação mais realista do desempenho do modelo.

```python
train_rus_model = apply_rus(
    train_base_final
)
```
